### Topdown

DO NOT use this, qn requires bottom up approach

```python
def unboundedKnapSack (index, space, weights, values):
    if index == len(values):
        return 0
    add = 0
    if weights[index] <= space:
        add = values[index] + unboundedKnapSack(index, space - weights[index], weights, values)
    
    no_add = unboundedKnapSack(index + 1, space, weights, values)

    return max(add, no_add)
```

### Bottom Up

```python
def unboundedKnapsack_bottom_up(C, weights, profits):
    # This builds the reference list
    # 1 is added to account for P(0)
    dp = [0] * (C + 1)
    
    n = len(weights)

    # Iterate through all capacities from 1 to C.
    for current_capacity in range(1, C + 1):
        # For each capacity, try to include each type of item.
        for i in range(n):
            item_weight = weights[i]
            item_profit = profits[i]
            
           
            if item_weight <= current_capacity:
                # Compare new profit and see if it is better than old value
                # Will auto add if its first value to be updated
                potential_profit = item_profit + dp[current_capacity - item_weight]
                dp[current_capacity] = max(dp[current_capacity], potential_profit)

    return dp[C]
```

What we are doing with the bottom up approach is to build on the solutions before it, i.e. get the best value from P(0) to P(C)

```python
W = [4, 6, 8]
P = [7, 6, 9]
space = 14

print(f"Top down approach: {unboundedKnapSack(0, 14, W, P)}")
print(f"Bottom up approach: {unboundedKnapsack_bottom_up(space, W, P)}")
```
